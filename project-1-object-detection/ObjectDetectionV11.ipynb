```python
import cv2
from ultralytics import YOLO
model_v11 = YOLO("yolo11x.pt")
image_name = "picasso-weeping-woman"
image_format = "jpg"
input_image_path = f"{image_name}.{image_format}"
output_image_path_v11 = f"{image_name}-v11.{image_format}"
results = model_v11.predict(input_image_path)
cv2.imwrite(output_image_path_v11, results[0].plot())
```
